```python
recalls = [0.1655, 0.2500, 0.4004, 0.5294]
precisions = [0.2445, 0.1938, 0.1330, 0.0925]
batches = ["@10", "@20", "@50", "@100"]

plt.figure(figsize=(8,5))
plt.plot(batches, recalls, marker='o', label="Recall", linewidth=2)
plt.plot(batches, precisions, marker='o', label="Precision", linewidth=2)
plt.xlabel("K")
plt.ylabel("Score")
plt.title("Recall and Precision vs K")

plt.legend()
plt.grid(True)

plt.show()
```
